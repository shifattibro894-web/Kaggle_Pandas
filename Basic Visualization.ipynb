# Bar Chart

```python
import matplotlib.pyplot as plt
subjects = ['Math', 'Physics', 'Chemistry', 'Biology',
'English', 'History']
results = [85, 78, 92, 88, 75, 80]
# Sorting for better readability (Ascending/Descending)
subjects, results = zip(*sorted(zip(subjects, results),
key=lambda x: x[1]))
plt.barh(subjects, results, color='skyblue')
plt.title('Class Results by Subject (Bar Chart)')
plt.xlabel('Marks')
plt.show()
```

```python
import matplotlib.pyplot as plt
branches=['Khilgaon','Gulshan','Banani','Dhanmondi','Bailey Road','Wari','Uttara']
sales=[100000,200000,400000,350000,500000,1000000,700000]
branches, sales = zip(*sorted(zip(branches, sales),
key=lambda x: x[1]))
plt.barh(branches,sales,color='brown')
plt.xlabel('Sales')
plt.ylabel('Branches')
plt.title('Regional Sales Analysis')
plt.show()
```

# Column Chart


```python
import matplotlib.pyplot as plt
branches=['Khilgaon','Gulshan','Banani','Dhanmondi','Bailey Road','Wari','Uttara']
sales=[100000,200000,400000,350000,500000,1000000,700000]
branches, sales = zip(*sorted(zip(branches, sales),
key=lambda x: x[1]))
plt.bar(branches,sales,color='brown')
plt.xlabel('Sales')
plt.ylabel('Branches')
plt.title('Regional Sales Analysis')
plt.show()
```

# Pie Chart

```python
import matplotlib.pyplot as plt
branches=['Khilgaon','Gulshan','Banani','Dhanmondi','Bailey Road','Wari','Uttara']
sales=[100000,200000,400000,350000,500000,1000000,700000]
colors = ['skyblue', 'yellowgreen']
plt.pie(sales, labels=branches, autopct='%1.1f%%',colors=colors,
startangle=140)
plt.title('Regional Sales Distribution')
plt.show()
```
