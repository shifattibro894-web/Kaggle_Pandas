# Section 2


Imputation

```python
import pandas as pd
file_path = '/content/business_analytics_lab_CT1_data.csv'
df=pd.read_csv(file_path)
df_3=df.copy()
df_3['Unit_Price']=df_3['Unit_Price'].fillna(df_3['Unit_Price'].mean())
df_3.isnull().sum()
```

Logical Validation

Finding negative numbers in 'Quantity'

```python
df4=df.copy()
negative_quantity_rows = df4[df4['Quantity'] < 0]


num_negative_quantities = len(negative_quantity_rows)

print(f"Number of rows with Quantity less than zero: {num_negative_quantities}")
```

Replacing with '0'

```python

df4.loc[df4['Quantity'] < 0, 'Quantity'] = 0


num_negative_quantities_after = len(df4[df4['Quantity'] < 0])
print(f"Number of rows with Quantity less than zero after replacement: {num_negative_quantities_after}")


display(df_3['Quantity'].describe())
```

Categorical Cleaning

```python
df_5=df4.copy()
df['Rating'].isnull().sum()
```

```python

df_5['Rating']= df_5['Rating'].fillna('Not Rated')
df_5['Rating'].isnull().sum()
```

# Section 3


Case Normalisation

```python

print("Product_Category unique values before normalization:")
display(df_5['Product_Category'].value_counts())

df_5['Product_Category'] = df_5['Product_Category'].str.lower()

print("\nProduct_Category unique values after normalization:")
display(df_5['Product_Category'].value_counts())
```

Trimming

```python
print("Customer_Name unique values before trimming:")
display(df_5['Customer_Name'].value_counts().head())

df_5['Customer_Name'] = df_5['Customer_Name'].str.strip()

print("\nCustomer_Name unique values after trimming:")
display(df_5['Customer_Name'].value_counts().head())
```

Standardization

```python
print("Unique 'State_Code' values (extracted from Contact_Location) before standardization:")
df_5['State_Code'] = df_5['Contact_Location'].apply(lambda x: x.split(', ')[-1] if ', ' in x else '')
display(df_5['State_Code'].value_counts())


df_5['State_Code'] = df_5['State_Code'].str.upper()

print("\nUnique 'State_Code' values after standardization to uppercase:")
display(df_5['State_Code'].value_counts())
```
